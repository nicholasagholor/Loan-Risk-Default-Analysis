```python
def create_age_group(age):
    if age == "<25":
        return "18-25"
    elif age == "25-34":
        return "26-35"
    elif age == "35-44":
        return "36-45"
    elif age == "45-54":
        return "46-55"
    elif age in ["55-64", "65-74", ">74"]:
         return "56+"

    
```

```python
##### Step 9: Create a Simple Customer Risk Classification

# Credit_Score < 670 = 1 point
# dtir1 > 43 = 1 point
# Income below 25th percentile = 1 point
# Loan amount above 75th percentile = 1 point
# Interest rate above 75th percentile = 1 point


income_low_threshold = df["income"].quantile(0.25)
loan_high_threshold = df["loan_amount"].quantile(0.75)
interest_high_threshold = df["rate_of_interest"].quantile(0.75)

print("Low Income Threshold:", income_low_threshold)
print("Large Loan Threshold:", loan_high_threshold)
print("High Interest Rate Threshold:", interest_high_threshold)
```

```python
df["Risk_Score"] = (
    (df["Credit_Score"] < 670).astype(int)
    + (df["dtir1"] > 43).astype(int)
    + (df["income"] < income_low_threshold).astype(int)
    + (df["loan_amount"] > loan_high_threshold).astype(int)
)
```

```python
# Create risk category

def classify_risk(score):
    if score <= 1:
        return "Low Risk"
    elif score <= 3:
        return "Moderate Risk"
    else:
        return "High Risk"


df["Risk_Category"] = df["Risk_Score"].apply(classify_risk)
print(df[["Risk_Score", "Risk_Category"]].head())
```

```python
# Numbers of customers in each category

print(df["Risk_Category"].value_counts())
```

```python
risk_default_analysis = (
    df.groupby("Risk_Category")
    .agg(
        Total_Loans=("Status", "count"),
        Total_Defaults=("Status", "sum"),
        Default_Rate=("Status", "mean")
    )
    .reset_index()
)

risk_default_analysis["Default_Rate"] = (
    risk_default_analysis["Default_Rate"] * 100
)

print(risk_default_analysis)
```

```python
# VISUALIZATION

fig = px.bar(
    risk_default_analysis,
    x="Risk_Category",
    y="Default_Rate",
    title="Default Rate by Customer Risk Category",
    labels={
        "Risk_Category": "Risk Category",
        "Default_Rate": "Default Rate (%)"
    }
)

fig.show()
```

```python
##### From the analysis, the risk classification shows a clear relationship with observed default rates: loans classified as High Risk have the highest default rate.
```
